Anonymize it using Presidio. Then use annonymized text for OpenAI prompt.

```python
text= f"""My name is Manami Arora. I lost my wallet recently.
Inside is some credit card with number 2345 1290 8000 3452.
If you find it, please call at 9870653450 or write an email here: arora_personal@gmail.com"""
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI

anonymizer= PresidioAnonymizer()
template= """Rewrite this text into official, short email:
{annomized_text}"""
prompt= PromptTemplate.from_template(template)
llm = ChatOpenAI( temperature=0)
chain={"annomized_text": annonymizer.anonymize} | prompt | llm
response= chain.invoke(text)
print(response.content)
```

```python
anonymized = anonymizer.anonymize(text)

print("ORIGINAL:")
print(text)

print("\nANONYMIZED:")
print(anonymized)
```

```python
annonymizer= PresidioAnonymizer(analyzed_fields=["PERSON","PHONE_NUMBER"])

annonymizer.anonymize("My name is Manami Arora. Please call at 9870653450 or write an email here: arora_personal@gmail.com")
```

```python
from presidio_analyzer import Pattern, PatternRecognizer

polish_phone_pattern = Pattern(
    name="polish_phone_number",
    regex=r"(?<!\d)(?:(?:\+48|0048)[ -]?)?\d{3}[ -]?\d{3}[ -]?\d{3}(?!\d)",
    score=0.7
)

polish_phone_recognizer = PatternRecognizer(
    supported_entity="POLISH_PHONE_NUMBER",
    patterns=[polish_phone_pattern]
)
```

```python
annonymizer.add_recognizer(polish_phone_recognizer)
```

```python
print(annonymizer.anonymize("My phone number is +48 123 456 789"))
print(annonymizer.anonymize("My phone number is 123 456 789"))
print(annonymizer.anonymize("My phone number is 0048 123 456 789"))
```

```python
from faker import Faker
fake= Faker(locale="pl_PL")

def fake_polish_phone_number(_=None):
  return fake.phone_number()

fake_polish_phone_number()
```

```python
from presidio_anonymizer import OperatorConfig

new_operators= {"POLISH_PHONE_NUMBER": OperatorConfig("custom", {"lambda": fake_polish_phone_number})}

annonymizer.add_operators(new_operators)
```

```python
annonymizer.anonymize("My phone number is 123 456 789")
```

Reversible Data Anonymization

```python
from langchain_experimental.data_anonymizer import PresidioReversibleAnonymizer
anonymizer= PresidioReversibleAnonymizer(
    analyzed_fields=["PERSON","PHONE_NUMBER","EMAIL_ADDRESS"],
    faker_seed=42
)

anonymizer.anonymize("My name is Manami Arora. Please call at 9870653450 or write an email here: arora_personal@gmail.com")
```

```python
fake_name="Alyssa Gonzalez"
fake_phone_number="(518)519-6001"
fake_email= "johnsonjoshua@example.org"

anonymized_text= f"""My name is {fake_name}. I lost my wallet recently.
Inside is some credit card with number 2345 1290 8000 3452.
If you find it, please call at {fake_phone_number} or write an email here: {fake_email}"""

print(anonymized_text)
```

```python
anonymizer= PresidioReversibleAnonymizer()
template= """Rewrite this text into official, short email:
{annomized_text}"""
prompt= PromptTemplate.from_template(template)
llm = ChatOpenAI( temperature=0)
chain={"annomized_text": annonymizer.anonymize} | prompt | llm
response= chain.invoke(text)
print(response.content)
```

```python
chain = chain | (lambda ai_message: anonymizer.deanonymize(ai_message.content))
response= chain.invoke(text)
print(response)
```
